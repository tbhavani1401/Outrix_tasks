```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("StudentsPerformance.csv")
print(df.head())
print(df.info())
print(df.describe())
```

```python
df.isnull().sum()
df.drop_duplicates(inplace=True)
```

```python
sns.histplot(df['math score'], kde=True, color='blue')
plt.show()

sns.countplot(data=df, x='gender', hue='gender', palette='Set2', legend=False)
plt.show()
```

```python
sns.boxplot(x='gender', y='math score', data=df)
plt.show()

numeric_df = df.select_dtypes(include=[np.number])   # select only numeric columns
sns.heatmap(numeric_df.corr(), annot=True, cmap='coolwarm')
plt.show()
```

```python
sns.pairplot(df[['math score','reading score','writing score']])
plt.show()
```

```python
sns.barplot(data=df, x='parental level of education', y='math score',
            hue='parental level of education', palette='mako', legend=False)
plt.xticks(rotation=45)
plt.show()
```
